# This is going to be used for a brief data analysis of the data

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns, datetime as dt, numpy as np, glob
parking_violation = pd.read_csv('csv_files/get_it_done_72_hour_violation_requests_datasd.csv')
graffiti_removal = pd.read_csv("csv_files/get_it_done_graffiti_requests_datasd.csv")
illegal_dumping = pd.read_csv('csv_files/get_it_done_illegal_dumping_requests_datasd.csv')
pot_hole = pd.read_csv("csv_files/get_it_done_pothole_requests_datasd.csv")
closed_2025 = pd.read_csv('csv_files/get_it_done_requests_closed_2025_datasd.csv')
left_open = pd.read_csv('csv_files/get_it_done_requests_open_datasd.csv')
```

## Parking — Overview
- Histogram showing the overall distribution of case ages.
- Summary statistics: mean and median case age shown below.

```python
parking_violation.sample(5)
parking_violation['date_requested'] = pd.to_datetime(parking_violation['date_requested'])
parking_violation["case_age_days"].plot(kind = "hist", bins = np.arange(0, 251, 5), ec='w', density = True)
plt.xlabel("Age of a case in days")
plt.ylabel("Density")
plt.title("Distribution of case age in days")
plt.show()
print(f"Mean age of a case is {parking_violation["case_age_days"].mean():.2f} days")
print(f"Median age of a case is {parking_violation["case_age_days"].median():.2f} days")


```
